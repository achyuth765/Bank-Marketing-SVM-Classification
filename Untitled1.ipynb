```python
import os
import time
import warnings
import json
import numpy as np
import pandas as pd
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split, StratifiedKFold, GridSearchCV, cross_val_score
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import (
    accuracy_score, balanced_accuracy_score, precision_score, recall_score,
    f1_score, roc_auc_score, confusion_matrix, classification_report
)

try:
    from imblearn.over_sampling import SMOTE
    HAS_IMBLEARN = True
except ImportError:
    HAS_IMBLEARN = False

warnings.filterwarnings("ignore")
RANDOM_STATE = 42
np.random.seed(RANDOM_STATE)

CSV_PATH = "bank-additional-full.csv"
OUT_DIR = "outputs"
os.makedirs(OUT_DIR, exist_ok=True)

TRAIN_SUBSAMPLE_SIZE = 4000   # see runtime note above
CV_FOLDS = 3                  # kept modest to bound grid-search runtime
MAX_ITER = 20000              # caps worst-case solver runtime for large-C /


REPORT_LINES = []


def log(msg=""):
    print(msg)
    REPORT_LINES.append(str(msg))


def section(title):
    log("\n" + "=" * 78)
    log(title)
    log("=" * 78)


def evaluate(y_true, y_pred, y_score, label):
    metrics = {
        "accuracy": accuracy_score(y_true, y_pred),
        "balanced_accuracy": balanced_accuracy_score(y_true, y_pred),
        "precision": precision_score(y_true, y_pred, zero_division=0),
        "recall": recall_score(y_true, y_pred, zero_division=0),
        "f1": f1_score(y_true, y_pred, zero_division=0),
        "roc_auc": roc_auc_score(y_true, y_score),
    }
    log(f"\n[{label}]")
    log(f"  Accuracy:          {metrics['accuracy']:.4f}")
    log(f"  Balanced accuracy: {metrics['balanced_accuracy']:.4f}")
    log(f"  Precision (yes):   {metrics['precision']:.4f}")
    log(f"  Recall (yes):      {metrics['recall']:.4f}")
    log(f"  F1 (yes):          {metrics['f1']:.4f}")
    log(f"  ROC-AUC:           {metrics['roc_auc']:.4f}")
    return metrics

```

## DATA LOADING AND PREPROCESSING

```python
df = pd.read_csv(CSV_PATH, sep=";")
log(f"Raw shape: {df.shape}")
log(f"Target distribution:\n{df['y'].value_counts().to_string()}")
minority_frac = df["y"].value_counts(normalize=True).min()
log(f"Minority class proportion: {minority_frac:.2%}  -> class imbalance is present.")

df = df.drop(columns=["duration"])
log("\nDropped 'duration': documented target leakage (only known AFTER the call")
log("outcome is already decided) - see UCI dataset documentation.")

log("\nMissing-value handling: 'unknown' entries in categorical columns are")
log("kept as their own explicit category (informative-missingness assumption),")
log("consistent with the dataset's own documentation.")

df["contacted_before"] = (df["pdays"] != 999).astype(int)
df["pdays_adj"] = df["pdays"].replace(999, -1)
df = df.drop(columns=["pdays"])
log("\nEngineered 'contacted_before' (binary) and 'pdays_adj' (999 -> -1) from")
log("'pdays' to avoid a scale-distorting sentinel value feeding into SVM's")
log("distance-based decision boundary.")

y = (df["y"] == "yes").astype(int)
X = df.drop(columns=["y"])

CATEGORICAL = X.select_dtypes(include="object").columns.tolist()
NUMERICAL = [c for c in X.columns if c not in CATEGORICAL]
log(f"\nCategorical features ({len(CATEGORICAL)}): {CATEGORICAL}")
log(f"Numerical features ({len(NUMERICAL)}): {NUMERICAL}")

preprocessor = ColumnTransformer([
    ("cat", OneHotEncoder(handle_unknown="ignore"), CATEGORICAL),
    ("num", StandardScaler(), NUMERICAL),
])
log("\nSVM is a distance/margin-based method, so ALL numeric features are")
log("standardised (zero mean, unit variance) and all categorical features are")
log("one-hot encoded - without this, features on larger numeric scales (e.g.")
log("nr.employed ~ thousands) would dominate the margin computation.")

# --- Splitting strategy ----------------------------------------------------------


X_train_pool, X_test, y_train_pool, y_test = train_test_split(
    X, y, test_size=0.30, stratify=y, random_state=RANDOM_STATE
)
X_train, _, y_train, _ = train_test_split(
    X_train_pool, y_train_pool, train_size=TRAIN_SUBSAMPLE_SIZE,
    stratify=y_train_pool, random_state=RANDOM_STATE
)
log(f"Training subsample used for all SVM fitting: {X_train.shape[0]} rows "
    f"(class balance {y_train.mean():.2%} positive)")
log(f"Held-out test set (full size, used for ALL reported test metrics): "
    f"{X_test.shape[0]} rows (class balance {y_test.mean():.2%} positive)")

Xt_train = preprocessor.fit_transform(X_train)
Xt_test = preprocessor.transform(X_test)
if hasattr(Xt_train, "toarray"):
    Xt_train = Xt_train.toarray()
    Xt_test = Xt_test.toarray()
log(f"Encoded feature dimensionality: {Xt_train.shape[1]}")
```


Data splitting approach
    A STRATIFIED train/test split (70:30) is used, preserving the ~11%/89%
    class ratio in both partitions - essential given the confirmed class
    imbalance, otherwise a random split risks a test set with too few
    minority-class ('yes') examples to evaluate reliably. Note: the raw data
    is ordered chronologically (May 2008-Nov 2010); a time-based split would
    be a more realistic deployment simulation (predicting future campaigns
    from past ones) and is mentioned here as a limitation of the random
    stratified split used for this analysis.


## Q1. LINEAR-KERNEL SVM (DEFAULT C = 1.0)

```python
t0 = time.time()
svm_linear_default = SVC(kernel="linear", C=1.0, probability=False, random_state=RANDOM_STATE, max_iter=MAX_ITER)
svm_linear_default.fit(Xt_train, y_train)
log(f"Fit time: {time.time() - t0:.2f}s")

pred_default = svm_linear_default.predict(Xt_test)
score_default = svm_linear_default.decision_function(Xt_test)  # signed distance to hyperplane
metrics_default = evaluate(y_test, pred_default, score_default, "Linear SVM, C=1.0 (default) - TEST SET")

cm = confusion_matrix(y_test, pred_default)
log("\nConfusion matrix (rows=actual, cols=predicted, order=[no, yes]):")
log(pd.DataFrame(cm, index=["no", "yes"], columns=["no", "yes"]).to_string())



fig, ax = plt.subplots(figsize=(4.5, 4))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", xticklabels=["no", "yes"],
            yticklabels=["no", "yes"], ax=ax)
ax.set_xlabel("Predicted"); ax.set_ylabel("Actual")
ax.set_title("Q1: Confusion Matrix - Linear SVM (C=1.0)")
plt.tight_layout()
plt.savefig(f"{OUT_DIR}/01_confusion_matrix_linear_default.png", dpi=120)
plt.close()

```


Metric choice rationale
    Given the ~89%/11% imbalance, plain accuracy is a poor primary metric (a
    model that always predicts 'no' would score ~89% accuracy while being
    useless). Balanced accuracy, F1, and ROC-AUC are reported alongside
    precision/recall for the minority 'yes' class as the metrics that
    actually reflect model usefulness for this task.


## Q2. LINEAR-KERNEL C SWEEP

```python
C_GRID = [0.001, 0.01, 0.1, 1, 10, 100]
cv = StratifiedKFold(n_splits=CV_FOLDS, shuffle=True, random_state=RANDOM_STATE)

cv_scores = []
for c in C_GRID:
    t0 = time.time()
    scores = cross_val_score(
        SVC(kernel="linear", C=c, random_state=RANDOM_STATE, max_iter=MAX_ITER),
        Xt_train, y_train, scoring="f1", cv=cv, n_jobs=-1
    )
    cv_scores.append(scores.mean())
    log(f"  C={c:<8} mean CV F1={scores.mean():.4f} (+/-{scores.std():.4f})  [{time.time()-t0:.1f}s]")

best_idx = int(np.argmax(cv_scores))
optimal_C = C_GRID[best_idx]
log(f"\nOptimal C (by 3-fold CV F1 on the training subsample): {optimal_C}")
log(f"Is it different from the default (C=1.0)? "
    f"{'YES' if optimal_C != 1.0 else 'NO'}")

fig, ax = plt.subplots(figsize=(6, 4.5))
ax.plot(C_GRID, cv_scores, marker="o")
ax.set_xscale("log")
ax.axvline(1.0, color="gray", linestyle="--", label="default C=1.0")
ax.axvline(optimal_C, color="red", linestyle="--", label=f"optimal C={optimal_C}")
ax.set_xlabel("C (log scale)"); ax.set_ylabel("Mean CV F1")
ax.set_title("Q2: Cross-validated F1 vs C (linear kernel)")
ax.legend()
plt.tight_layout()
plt.savefig(f"{OUT_DIR}/02_C_sweep_cv_score.png", dpi=120)
plt.close()

# Fit final default vs optimal models on the full training subsample, evaluate on test
svm_linear_optimal = SVC(kernel="linear", C=optimal_C, random_state=RANDOM_STATE, max_iter=MAX_ITER)
svm_linear_optimal.fit(Xt_train, y_train)
pred_optimal = svm_linear_optimal.predict(Xt_test)
score_optimal = svm_linear_optimal.decision_function(Xt_test)
metrics_optimal = evaluate(y_test, pred_optimal, score_optimal, f"Linear SVM, C={optimal_C} (optimal) - TEST SET")

log("\nDefault vs optimal C - test set comparison:")
comp_df = pd.DataFrame({
    f"C=1.0 (default)": metrics_default,
    f"C={optimal_C} (optimal)": metrics_optimal,
}).T
log(comp_df.round(4).to_string())

# Margin width: for a linear-kernel SVM, margin = 2 / ||w||_2
w_default = svm_linear_default.coef_.ravel()
w_optimal = svm_linear_optimal.coef_.ravel()
margin_default = 2.0 / np.linalg.norm(w_default)
margin_optimal = 2.0 / np.linalg.norm(w_optimal)
log(f"\nMargin width (2 / ||w||): default C=1.0 -> {margin_default:.5f} | "
    f"optimal C={optimal_C} -> {margin_optimal:.5f}")
log(f"Support vectors: default C=1.0 -> {svm_linear_default.support_vectors_.shape[0]} | "
    f"optimal C={optimal_C} -> {svm_linear_optimal.support_vectors_.shape[0]}")

fig, ax = plt.subplots(figsize=(5, 4.5))
bars = ax.bar(["C=1.0 (default)", f"C={optimal_C} (optimal)"],
              [margin_default, margin_optimal], color=["#1f77b4", "#ff7f0e"])
ax.set_ylabel("Margin width (2 / ||w||)")
ax.set_title("Q2: Margin width, default vs optimal C")
for b, v in zip(bars, [margin_default, margin_optimal]):
    ax.text(b.get_x() + b.get_width() / 2, v, f"{v:.4f}", ha="center", va="bottom")
plt.tight_layout()
plt.savefig(f"{OUT_DIR}/03_margin_comparison.png", dpi=120)
plt.close()


```


How the margin changes with C
    C controls the trade-off between margin width and training misclassification
    penalty: a SMALL C tolerates more margin violations in exchange for a WIDER
    margin (more regularisation, simpler boundary); a LARGE C penalises
    violations heavily, producing a NARROWER margin that fits the training
    data more tightly (less regularisation, higher variance).
    Here, moving from C=1.0 to the optimal C={optimal_C}, the margin width
    changed from {margin_default:.5f} to {margin_optimal:.5f}
    ({"narrowing" if margin_optimal < margin_default else "widening"} the
    margin), which is consistent with {"an increase" if optimal_C > 1.0 else
    "a decrease" if optimal_C < 1.0 else "no change"} in C: {"a larger C "
    "prioritises fitting the (heavily overlapping, imbalanced) training data "
    "more closely at the cost of a tighter margin." if optimal_C > 1.0 else
    "a smaller C tolerates more overlap between the two classes, favouring a "
    "wider, more conservative margin - sensible given the substantial class "
    "overlap typical of behavioural marketing data." if optimal_C < 1.0 else
    "the default already balanced margin width and error tolerance well for "
    "this data."}

## Q3. RBF AND POLYNOMIAL KERNEL TUNING

```python
rbf_grid = {"C": [0.1, 1, 10, 100], "gamma": ["scale", 0.01, 0.1, 1]}
t0 = time.time()
rbf_search = GridSearchCV(SVC(kernel="rbf", random_state=RANDOM_STATE, max_iter=MAX_ITER), rbf_grid,
                           scoring="f1", cv=cv, n_jobs=-1)
rbf_search.fit(Xt_train, y_train)
log(f"RBF grid search time: {time.time()-t0:.1f}s")
log(f"RBF best params: {rbf_search.best_params_}  |  best CV F1: {rbf_search.best_score_:.4f}")

poly_grid = {"C": [0.1, 1, 10], "degree": [2, 3, 4], "gamma": ["scale"]}
t0 = time.time()
poly_search = GridSearchCV(SVC(kernel="poly", random_state=RANDOM_STATE, max_iter=MAX_ITER), poly_grid,
                            scoring="f1", cv=cv, n_jobs=-1)
poly_search.fit(Xt_train, y_train)
log(f"Poly grid search time: {time.time()-t0:.1f}s")
log(f"Poly best params: {poly_search.best_params_}  |  best CV F1: {poly_search.best_score_:.4f}")

rbf_best = rbf_search.best_estimator_
poly_best = poly_search.best_estimator_

pred_rbf = rbf_best.predict(Xt_test)
score_rbf = rbf_best.decision_function(Xt_test)
metrics_rbf = evaluate(y_test, pred_rbf, score_rbf, f"RBF SVM, best params {rbf_search.best_params_} - TEST SET")

pred_poly = poly_best.predict(Xt_test)
score_poly = poly_best.decision_function(Xt_test)
metrics_poly = evaluate(y_test, pred_poly, score_poly, f"Poly SVM, best params {poly_search.best_params_} - TEST SET")

log("\nAll-kernel comparison (test set):")
all_kernel_df = pd.DataFrame({
    "Linear (optimal C)": metrics_optimal,
    "RBF (tuned)": metrics_rbf,
    "Poly (tuned)": metrics_poly,
}).T
log(all_kernel_df.round(4).to_string())

kernel_models = {
    "Linear (optimal C)": svm_linear_optimal,
    "RBF (tuned)": rbf_best,
    "Poly (tuned)": poly_best,
}
kernel_f1 = {name: all_kernel_df.loc[name, "f1"] for name in kernel_models}
best_kernel_name = max(kernel_f1, key=kernel_f1.get)
best_model = kernel_models[best_kernel_name]
log(f"\nOverall best kernel/hyperparameter setting (by test F1): {best_kernel_name}")

fig, ax = plt.subplots(figsize=(7, 4.5))
all_kernel_df[["f1", "roc_auc", "balanced_accuracy"]].plot(kind="bar", ax=ax)
ax.set_title("Q3: Kernel comparison on test set")
ax.set_ylabel("Score")
plt.xticks(rotation=15)
plt.tight_layout()
plt.savefig(f"{OUT_DIR}/04_kernel_comparison.png", dpi=120)
plt.close()
```

## Q4. SUPPORT VECTORS - BEST MODEL

```python
n_sv_total = best_model.support_vectors_.shape[0]
n_sv_per_class = dict(zip(["no", "yes"], best_model.n_support_.tolist()))
log(f"Best model: {best_kernel_name}")
log(f"Total support vectors: {n_sv_total} out of {Xt_train.shape[0]} training rows "
    f"({n_sv_total / Xt_train.shape[0]:.1%})")
log(f"Support vectors per class: {n_sv_per_class}")

```


What the support vectors indicate
    Support vectors are the training points that lie exactly on, or on the
    wrong side of, the margin - i.e. the points closest to (or violating) the
    decision boundary. They are the ONLY training points that influence the
    final boundary: every other point could be removed without changing the
    fitted hyperplane. A large support-vector count relative to the training
    set size (as seen here) indicates substantial overlap between the 'yes'
    and 'no' classes - many clients with similar profiles end up on both
    sides of the outcome - which matches the intuitively "noisy" nature of
    behavioural marketing response data (two clients with near-identical
    demographic/economic features can still differ in whether they subscribe).
    This also foreshadows the SVM's scalability problem in production:
    prediction cost and memory both scale with the number of support
    vectors, and a high proportion of support vectors is itself a symptom of
    a difficult, low-margin classification problem for this kernel/setting.

## Q5. CLASS IMBALANCE HANDLING FOR SVM

```python
log("""
Method chosen: class_weight='balanced'
    Rather than resampling the data (which changes the geometry the SVM
    solver sees and, for a small training subsample, discards or duplicates
    already-scarce information), class_weight='balanced' re-weights the C
    penalty per class inversely proportional to class frequency
    (C_class = C * n_samples / (n_classes * n_samples_in_class)). This is a
    method built directly into the SVM optimisation objective: minority-class
    ('yes') margin violations become proportionally more costly, pushing the
    boundary to be less biased toward the majority class - without altering
    the underlying training distribution, adding synthetic points, or the
    extra computational overhead of resampling. It is the most natural,
    SVM-native way to address imbalance and is applied here using the same
    best kernel/hyperparameters identified in Q3.
""")

best_params = best_model.get_params()
weighted_params = {k: v for k, v in best_params.items()
                    if k in ("kernel", "C", "gamma", "degree")}
weighted_model = SVC(class_weight="balanced", random_state=RANDOM_STATE, max_iter=MAX_ITER, **weighted_params)
weighted_model.fit(Xt_train, y_train)
pred_weighted = weighted_model.predict(Xt_test)
score_weighted = weighted_model.decision_function(Xt_test)
metrics_weighted = evaluate(y_test, pred_weighted, score_weighted,
                             f"{best_kernel_name} + class_weight='balanced' - TEST SET")

log("\nBefore vs after class-weighting (test set):")
imbalance_df = pd.DataFrame({
    f"{best_kernel_name} (unweighted)": all_kernel_df.loc[best_kernel_name],
    f"{best_kernel_name} (class_weight='balanced')": metrics_weighted,
}).T
log(imbalance_df.round(4).to_string())

# Optional secondary comparison: SMOTE oversampling (if available)
if HAS_IMBLEARN:
    sm = SMOTE(random_state=RANDOM_STATE)
    Xt_train_sm, y_train_sm = sm.fit_resample(Xt_train, y_train)
    smote_model = SVC(random_state=RANDOM_STATE, max_iter=MAX_ITER, **weighted_params)
    smote_model.fit(Xt_train_sm, y_train_sm)
    pred_smote = smote_model.predict(Xt_test)
    score_smote = smote_model.decision_function(Xt_test)
    metrics_smote = evaluate(y_test, pred_smote, score_smote,
                              f"{best_kernel_name} + SMOTE (secondary comparison) - TEST SET")
    imbalance_df.loc[f"{best_kernel_name} (SMOTE, for comparison)"] = pd.Series(metrics_smote)
    log("\nFull imbalance-method comparison (test set):")
    log(imbalance_df.round(4).to_string())

fig, ax = plt.subplots(figsize=(7, 4.5))
imbalance_df[["recall", "f1", "balanced_accuracy"]].plot(kind="bar", ax=ax)
ax.set_title("Q5: Imbalance handling - recall/F1/balanced accuracy")
plt.xticks(rotation=15, ha="right")
plt.tight_layout()
plt.savefig(f"{OUT_DIR}/05_imbalance_comparison.png", dpi=120)
plt.close()

```


Interpretation
    class_weight='balanced' is expected to raise recall on the minority
    'yes' class substantially (fewer subscribing clients missed), typically
    at some cost to precision and overall accuracy - a deliberate,
    justified trade given that missing a genuine subscriber is usually more
    costly to a bank's marketing objective than an extra call to a client
    who declines. Balanced accuracy and F1 provide the fairer summary of
    whether the trade was worthwhile; compare the table above directly.
